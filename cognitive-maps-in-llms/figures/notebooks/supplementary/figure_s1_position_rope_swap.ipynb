```python
%config InlineBackend.figure_format = 'retina'
%load_ext autoreload
%autoreload 2

from pathlib import Path

import matplotlib.pyplot as plt
from IPython.display import display

from cognitive_maps_in_llms.analysis import RunResults, plot_mosaic_series
```

```python
REPO_ROOT = next(
    path for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (path / 'src' / 'cognitive_maps_in_llms').is_dir()
)

# SET INPUT AND OUTPUT PATHS

# ------------------------------------------------------------------------------
RESULTS_ROOT = REPO_ROOT / 'results/final_run_2026-09-30'
FIGURES_DIR  = REPO_ROOT / 'figures/generated/supplementary/figure_s1'
SAVE_FIGURES = True
# ------------------------------------------------------------------------------

if SAVE_FIGURES:
    FIGURES_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# DEFINE HELPER TO MAKE A SINGLE POSITION/ROPE SWAP PLOT

MODEL_PRESET = 'gemma-27b'
PROMPT_PRESET = 'ABC_family_tree_yes_no'
EXPERIMENT = 'exp_s1_positions_rope'
CONTROL_CONDITION = 'CTRL'
SWAP_CONDITION = 'SWAP'
RESULTS = RunResults(results_root=RESULTS_ROOT)

def plot_figure_s1_query(query: str, title: str, filename: str):
    control = RESULTS.load_series(
        model_preset=MODEL_PRESET,
        prompt_preset=PROMPT_PRESET,
        experiment=EXPERIMENT,
        query=query,
        condition=CONTROL_CONDITION,
    )
    swap = RESULTS.load_series(
        model_preset=MODEL_PRESET,
        prompt_preset=PROMPT_PRESET,
        experiment=EXPERIMENT,
        query=query,
        condition=SWAP_CONDITION,
    )

    plot = plot_mosaic_series(
        {'Control': control, 'Mosaic': swap},
        value_kind='probability_ratio',
        title=title,
        legend=True,
    )
    if SAVE_FIGURES:
        plot.save(FIGURES_DIR / filename)
    return plot
```

```python
FIGURE_SPECS = (
    ('A --daughter--> B?', "Is ‹A›'s daughter ‹B›?", 'is_As_daughter_B.svg'),
    ('B --daughter--> A?', "Is ‹B›'s daughter ‹A›?", 'is_Bs_daughter_A.svg'),
    ('A --mother--> B?', "Is ‹A›'s mother ‹B›?", 'is_As_mother_B.svg'),
    ('B --mother--> A?', "Is ‹B›'s mother ‹A›?", 'is_Bs_mother_A.svg'),
)

for query, title, filename in FIGURE_SPECS:
    plot = plot_figure_s1_query(query, title, filename)
    display(plot.fig)
    plt.close(plot.fig)
```
